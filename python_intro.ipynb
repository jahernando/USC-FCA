### 2.1 Some rules, from a physicist

1. Beautiful is better than ugly.
2. Keep it simple. Simple is better than complex.
3. Readability counts.
4. Document your code: docstrings, and comments where the code cannot speak for itself.
5. Test your code, or it may not do what you think it does.
6. Flat is better than nested.
7. Code must be modular: big things are made of small parts. Reuse them.
8. If the implementation is hard to explain, it is a bad idea.
9. Do not over-engineer: code what you need.
10. If a pattern repeats, there is an abstraction behind it: exploit it.

In short: **think before you write**.

## 3. Good manners: PEP 8

The style guide for Python is [PEP 8](https://peps.python.org/pep-0008/). Some rules:

1. Lines of at most 79 characters.
2. Spaces, not tabs; 4 spaces per indentation level.
3. Blank lines between functions and classes.
4. Imports at the top of the module, one per line. Never `from module import *`.
5. A space after commas and around operators: `a = b + c`, not `a=b+c`.
6. Meaningful comments, in English.
7. Names: `ClassName` for classes, `UPPER_CASE` for constants, `lower_case` for functions,
   variables and modules.
8. Meaningful names; words separated by `_`. Short, lowercase names for modules.
9. A leading underscore, `_name`, means *private* by convention: for internal use.
10. Do not hide errors: raise exceptions, and catch only those you can handle.

    for x in vector:

<details>
<summary><b>Solution</b></summary>

The second one. Same algorithm, but:

* the name says what it does (`norm`), and so do the arguments (`vector`, not `l`,
  which is easy to confuse with `1`);
* it has a docstring;
* it iterates over the elements, not over the indices;
* it follows PEP 8 (spaces, one statement per line).

Code is read many more times than it is written.

import math

SPEED_OF_LIGHT = 299792458.  # m/s


def kinetic_energy(mass, speed, c=SPEED_OF_LIGHT):
    """Relativistic kinetic energy (J); mass in kg, speed in m/s."""
    gamma = 1. / math.sqrt(1. - (speed / c) ** 2)
    return (gamma - 1.) * mass * c ** 2


for m, v in [(1., 0.), (9.109e-31, 1e8), (1.673e-27, 2.9e8)]:

## 4. Programming with AI

You will use AI assistants to write code. Use them, but so that you still learn:

1. Try first, and decide which **tests** your code must pass.
2. Check that your code passes them.
3. Ask the AI to review it: *what is wrong or incomplete?*
4. Ask it for an alternative version: code **and** tests.
5. **Cross**: does its code pass your tests? Does yours pass its tests?
6. Compare both versions.
7. Redo yours if needed.

The rule: you may use AI, but you **declare** it, and you never hand in code that you
cannot explain. In the exams there is no computer.

<details>
<summary><b>[i] Cheat sheet — Before you write a function</b></summary>

* What does it take, what does it return? Write the docstring first.
* Which inputs must it handle? Which are wrong? Write them as `assert`s.
* Choose names that make comments unnecessary.
* Is it already in the standard library, numpy or scipy? Then do not write it.
