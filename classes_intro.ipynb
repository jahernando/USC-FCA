## 1. Class and object

A **class** (or **type**) is the definition of a kind of thing: what data it has and what
can be done with it. Think of *a car* in general.

An **object** (or **instance**) is one concrete thing of that kind. Think of *my* car.

<img src="./img/classes_car.png" width="350">

A class defines:

* **attributes**: the data of each object. A complex number has a real and an imaginary
  part;
* **methods**: the functions that act on the object. A complex number can be conjugated,
  added, multiplied…

`int`, `float`, `str`, `list`, `dict`, `complex`, `numpy.ndarray`: all are classes, and
every value in Python is an object of some class.

## 2. The complex class

### 2.1 Construction and type

An object is created by calling the class, the **constructor**. Some types also have a
literal syntax:

````python
x = complex(1., 2.)
y = 1 + 2j

print(x, y, x == y)
print(type(x), isinstance(x, complex))
````

### 2.2 Getting help

`help(complex)` shows the documentation of the class: its constructor, attributes and
methods. `dir(x)` lists the names an object has.

````python
help(complex.conjugate)
````

````python
print([name for name in dir(x) if not name.startswith('_')])
````

### 2.3 Attributes and methods

Both are accessed with a dot. An **attribute** is a value; a **method** is called, with
`()`:

````python
x = 3 - 4j
print(x.real, x.imag)        # attributes
print(x.conjugate())         # method
````

### 2.4 A method is a function of the class

`x.conjugate()` is a shortcut for `complex.conjugate(x)`: the function `conjugate`,
defined inside the class, applied to the object `x`.

````python
print(x.conjugate())
print(complex.conjugate(x))
````

The object is the **first argument** of the method. When we write our own classes, this
first argument will have a name: `self`.

## 3. Special methods

The methods whose names start and end with two underscores, `__name__`, are **special**:
Python calls them for you when you use an operator or a built-in function.

````python
x, y = 1 + 1j, 2 - 3j

print(x + y, x.__add__(y))
print(abs(x), x.__abs__())
print(str(x), x.__str__())
````

| You write | Python calls |
|:--|:--|
| `x + y`, `x - y`, `x * y`, `x / y` | `x.__add__(y)`, `__sub__`, `__mul__`, `__truediv__` |
| `x ** y`, `x @ y`, `-x` | `__pow__`, `__matmul__`, `__neg__` |
| `x == y`, `x < y` | `__eq__`, `__lt__` |
| `abs(x)`, `len(x)`, `str(x)`, `repr(x)` | `__abs__`, `__len__`, `__str__`, `__repr__` |
| `x[i]`, `x[i] = v` | `__getitem__`, `__setitem__` |
| `for v in x` | `__iter__` |
| `ClassName(...)` | `__new__`, `__init__` |

Not all of them apply to `complex` (it has no `<`, `len` or `[i]`). This is why a class
of ours can have `+` and `print`: we only need to write those methods. If `a.__add__(b)`
cannot handle `b`, Python tries `b.__radd__(a)`.

<p class="admonition-title"><b>[=] Commit — The same +</b></p>

What does `add` return for each pair? Why does the same function work for so many types?
Does any call fail?

```python
def add(a, b):
    return a + b

add(1, 2)
```

</div>

````python
def add(a, b):
    return a + b

print(add(1, 2), add(1 + 2j, 3.), add('ab', 'cd'), add([1, 2], [3]))
add([1, 2], 3)
````

<details>
<summary><b>Solution</b></summary>

`3`, `(4+2j)`, `'abcd'`, `[1, 2, 3]`, and a `TypeError` for the last one.

`add` does not know the types: it calls `a.__add__(b)`, and each class decides what `+`
means: numbers add, strings and lists concatenate. A list does not know how to add an
`int`, so the last call fails. This is **duck typing**, polymorphism the Python way (recall
`simulate` in [Why classes?](./fca_classes.ipynb)): what matters is what an object *can
do*, not what it *is*.

</details>

## 4. Names and objects

A variable is a **name** that refers to an object. An assignment `y = x` does not copy the
object: it gives it a second name.

<div class="admonition important">
<p class="admonition-title"><b>[=] Commit — One list or two?</b></p>

**Decide** what the two `print` show.

```python

print(y)
print(z)
```

</div>

````python
x = [1, 2]
y = x
z = list(x)
x[0] = 99
print(y)
print(z)
print(y is x, z is x)
````

<details>
<summary><b>Solution</b></summary>

`[99, 2]` and `[1, 2]`. `y = x` gives a second name to the **same** list, so the change
is seen through both names. `list(x)` builds a **new** list, a copy. `is` checks
whether two names refer to the same object.

With numbers and strings this never surprises you, because they are **immutable**: no
operation changes them, it always creates a new object. Lists, dicts, sets, numpy arrays
and most of the objects we will write are **mutable**.

</details>

## 5. Summary

* Every value is an object of a class; a class bundles data (attributes) and behaviour
  (methods).
* Operators and built-ins call special methods.
* A variable is a name: `y = x` does not copy.

<details>
<summary><b>[i] Cheat sheet — Using classes</b></summary>

| | |
|:--|:--|
| class / type | the definition: attributes and methods |
| object / instance | one concrete value of a class |
| `C(args)` | constructor: creates an object (`__init__`) |
| `type(x)`, `isinstance(x, C)` | the class of an object |
| `help(C)`, `dir(x)` | documentation, names |
| `x.attr`, `x.method()` | attribute, method |
| `x.method()` = `C.method(x)` | the object is the first argument |
| `x + y` → `x.__add__(y)` | operators and built-ins call special methods |
| `str` / `repr` | for the user / for the programmer |
| `y = x` | a second name, not a copy; `is` checks identity |

</details>
