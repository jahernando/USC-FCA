### 1 &middot; Get a local copy

Any of these:

* In the repository page, under the **Code** tab, *Download ZIP*, and unzip it.

* In a terminal, clone it:

```bash
git clone https://github.com/jahernando/USC-FCA.git
```

* If you have a GitHub account, **fork** it to your own area and then clone your fork.
  This is the option to choose if you want to keep your own notes and exercises under
  version control.

### 2 &middot; Create the environment

The repository ships an `environment.yml` with everything the course needs.
From the directory where you cloned it:

```bash
conda env create -f environment.yml
conda activate fca
```

That gives you Python 3.13, JupyterLab, numpy, scipy, matplotlib, pandas, sympy, pytest
and the slides extension. You only do this once; afterwards, just `conda activate fca`.

### 3 &middot; Open the notebooks

From the same directory, with the environment active:

```bash
jupyter lab
```

Your browser opens with the file list. Start with [`index.ipynb`](./index.ipynb).

### The notebooks are also the slides

The same notebooks are what is projected in class. The environment includes
**`jupyterlab_rise`**, which turns a notebook into a presentation.

* Open a notebook and press **Alt-R** to enter presentation mode, and again to leave it.
* Space and arrows move through the slides; the cells still run.

Note that `jupyterlab_rise` is *not* part of Jupyter: it is a separate package, already
listed in `environment.yml`. The older extension, simply called `rise`, only worked with
the classic Notebook and does **not** work with current versions.

### If you cannot install anything

You can open the notebooks in [Google Colab](https://colab.research.google.com) from the
GitHub repository, with nothing installed locally.

Use it only as a fallback: Colab does not run the slides, the UNIX and git notebooks assume
a real shell of your own, and anything you change is lost unless you save it back yourself.
For the course, install the environment.
